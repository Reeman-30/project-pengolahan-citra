```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
from google.colab.patches import cv2_imshow
```

```python
img = cv2.imread('/content/drive/MyDrive/Footage/boneka.jpg', 0)

row, column = img.shape

img1 = np.zeros((row,column), dtype=np.uint8)
img2 = np.zeros((row,column,3), dtype=np.uint8)

min_range = 10
max_range = 60

for i in range (row):
  for j in range (column):
    if (img[i,j] >= min_range) and (img[i,j] <= max_range):
      img1[i,j] = 255
      img2[i,j] = [255,0,0]
    else:
      img1[i,j] = 0
      img2[i,j] = [0,0,255]

plt.figure(figsize=(12,6))
plt.subplot(131), plt.imshow(cv2.cvtColor(img, cv2.COLOR_GRAY2BGR))
plt.title('Citra Asli')
plt.axis('off')

plt.subplot(132), plt.imshow(cv2.cvtColor(img1, cv2.COLOR_BGR2RGB))
plt.title('Citra Intensity Slicing')
plt.axis('off')

plt.subplot(133), plt.imshow(cv2.cvtColor(img2, cv2.COLOR_BGR2RGB))
plt.title('Citra Slicing with color')
plt.axis('off')

plt.show()
```

```python
def bgr2cmyk(img):
    cmyk = np.zeros(img.shape, dtype=np.float32)
    cmyk[:, :, 0] = 1 - img[:, :, 2]
    cmyk[:, :, 1] = 1 - img[:, :, 1]
    cmyk[:, :, 2] = 1 - img[:, :, 0]
    k = np.min(cmyk, axis=2)
    return cmyk, k

cmyk, k = bgr2cmyk(img)

cv2_imshow(cmyk)

fig, axes = plt.subplots(1, 4, figsize=(15, 5))

axes[0].imshow(cmyk[:, :, 0], cmap='gray')
axes[0].set_title('Cyan')

axes[1].imshow(cmyk[:, :, 1], cmap='gray')
axes[1].set_title('Magenta')

axes[2].imshow(cmyk[:, :, 2], cmap='gray')
axes[2].set_title('Yellow')

axes[3].imshow(k, cmap='gray')
axes[3].set_title('Black')

plt.show()
```

```python
def color_complement(rgb_color):
    if len(rgb_color) != 3 or any(value < 0 or value > 255 for value in rgb_color):
        raise ValueError("Input harus merupakan warna RGB yang valid dalam rentang [0, 255] untuk setiap saluran")

    complement = [255 - value for value in rgb_color]

    return complement

input_color = [201, 0, 0]

complement_color = color_complement(input_color)

print("Input Color:", input_color)
print("Color Complement:", complement_color)
```

```python
img = cv2.imread("/content/drive/MyDrive/Footage/monarch.tif")

def gray_world_balance(image):
    # Convert image to float to prevent overflow
    image_float = image.astype(np.float64)

    # Calculate the average value of each channel
    b_mean, g_mean, r_mean = np.mean(image_float[:, :, 0]), np.mean(image_float[:, :, 1]), np.mean(image_float[:, :, 2])

    # Calculate the overall gray mean
    gray_mean = (b_mean + g_mean + r_mean) / 3.0

    # Scale each channel
    image_float[:, :, 0] *= (gray_mean / b_mean)
    image_float[:, :, 1] *= (gray_mean / g_mean)
    image_float[:, :, 2] *= (gray_mean / r_mean)

    # Clip values to [0, 255] and convert back to uint8
    return np.clip(image_float, 0, 255).astype(np.uint8)

corrected_img = gray_world_balance(img)

cv2_imshow(np.hstack((img, corrected_img)))
```

```python
img = cv2.imread("/content/drive/MyDrive/Footage/peppers3.tif")

hsi_img = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

lower_hue = 30  # Batas bawah untuk hue (contoh green)
upper_hue = 90  # Batas atas untuk hue (contoh blue-green)

mask = cv2.inRange(hsi_img, (lower_hue, 0, 0), (upper_hue, 255, 255))

segmented_img = cv2.bitwise_and(img, img, mask=mask)

cv2_imshow(np.hstack((img, segmented_img)))
```

```python
num_colors = 16
reshaped_img = img.reshape((-1, 3))
reshaped_img = np.float32(reshaped_img)
criteria = (cv2.TERM_CRITERIA_EPS + cv2.TERM_CRITERIA_MAX_ITER, 100, 0.2)
_, labels, centers = cv2.kmeans(reshaped_img, num_colors, None, criteria, 10,
  cv2.KMEANS_RANDOM_CENTERS)
centers = np.uint8(centers)
compressed_img = centers[labels.flatten()].reshape(img.shape)

cv2_imshow(np.hstack((img, compressed_img)))
```
